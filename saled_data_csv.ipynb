```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

```python
dft.drop(columns=['Status','unnamed1'],inplace=True)
```

```python
dft
```

```python
dft.dropna(inplace=True)
```

```python
dft
```

```python
dft.rename(columns={'Marital_Status':'Marital'})
```

```python
dft['Marital_Status']=dft['Marital_Status'].replace({0:'single',1:'married'})
```

```python
dft
```

```python
dft['Gender']=dft['Gender'].replace({'F':'Female','M':'Male'})
```

```python
dft
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(facecolor='orange',figsize=(10,10))
plt.suptitle('Sales Data',color='red',fontsize=20)
plt.subplot(2,2,1)
plt.pie(dft['Gender'].value_counts(),labels=dft['Gender'].value_counts().index,autopct='%1.1f%%')
plt.title('Gender',color='blue',fontsize=15)
plt.gca().set_facecolor('white')

plt.subplot(2,2,2)
count=sns.countplot(x='Gender',data=dft)
plt.title('Gender ratio',color='pink',fontsize=15)
plt.gca().set_facecolor('green')
plt.xlabel('Gender',color='violet',fontsize=15)
plt.ylabel('Count',color='purple',fontsize=15)
plt.subplot(2,2,3)
count=sns.countplot(x='Marital_Status',data=dft)
plt.title('Marital Status ratio',color='red',fontsize=15)
plt.gca().set_facecolor('pink')
for i in count.containers:
  count.bar_label(i)
plt.xlabel('Marital Status',color='yellow',fontsize=15)
plt.ylabel('Count',color='black',fontsize=15)
plt.subplot(2,2,4)
plt.pie(dft['Marital_Status'].value_counts(),labels=dft['Marital_Status'].value_counts().index,autopct='%1.1f%%')
plt.title('Maritalratio',color='black',fontsize=15)
plt.gca().set_facecolor('pink')
plt.xlabel('Marital Status',color='black',fontsize=15)
plt.ylabel('Count',color='purple',fontsize=15)
plt.show()
```
